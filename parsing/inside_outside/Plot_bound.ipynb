```python

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.patches as mpatches

plt.rcParams['legend.fontsize'] = 14  # or any size you prefer

def weighted_avg_by(rows, group_col, val_col=3, count_col=4):
    keys = np.unique(rows[:, group_col].astype(int))
    avg = np.empty(len(keys))
    for i, k in enumerate(keys):
        sel = rows[:, group_col].astype(int) == k
        counts = rows[sel, count_col]
        vals = rows[sel, val_col]
        avg[i] = np.dot(counts, vals) / counts.sum()
    return keys, avg

# --- Tree parameters ---
L = 3
b = 3
n_leaves = b**L  # 27

# Build node x-positions: leaves at 0..26, internal nodes at centre of children
def build_tree_x(L, b):
    pos = {}
    for i in range(b**L):
        pos[(L, i)] = float(i)
    for level in range(L - 1, -1, -1):
        for i in range(b**level):
            pos[(level, i)] = np.mean([pos[(level + 1, i * b + j)] for j in range(b)])
    return pos

pos = build_tree_x(L, b)

# One colour per level-1 subtree
palette = ['#e41a1c', '#377eb8', '#4daf4a']

def node_color(level, idx):
    """Colour a node/edge by its level-1 ancestor."""
    if level == 0:
        return '#555555'
    return palette[(idx // b**(level - 1)) % len(palette)]

# --- Figure ---
fig = plt.figure(figsize=(10, 6.5))
gs = gridspec.GridSpec(2, 1, height_ratios=[1, 2.5], hspace=0.06)
ax_tree = fig.add_subplot(gs[0])
ax_main = fig.add_subplot(gs[1], sharex=ax_tree)

# ---- Draw tree ----
for level in range(L):
    for i in range(b**level):
        px, py = pos[(level, i)], float(L - level)
        for j in range(b):
            ci = i * b + j
            cx, cy = pos[(level + 1, ci)], float(L - level - 1)
            col = node_color(level + 1, ci)
            ax_tree.plot([px, cx], [py, cy], '-', color=col, lw=1.8, solid_capstyle='round', zorder=2)
            ax_tree.plot(cx, cy, 'o', color=col, markersize=4.5, zorder=3)

# Root node
ax_tree.plot(pos[(0, 0)], float(L), 'o', color='#555555', markersize=7, zorder=3)

# Level labels on the right
level_labels = {0: 'root', 1: r'$\ell=2$', 2: r'$\ell=1$', 3: 'leaves'}
for level in range(L + 1):
    ax_tree.text(n_leaves - 0.2, float(L - level), level_labels[level],
                 ha='left', va='center', fontsize=7.5, color='#444444')

# Thin vertical dashed lines at level-1 boundaries (every 9 tokens), shared with main plot
for k in range(1, b):
    x_split = k * (n_leaves // b) - 0.5
    ax_tree.axvline(x_split, color='#aaaaaa', lw=0.8, ls='--', zorder=1)
    ax_main.axvline(x_split, color='#aaaaaa', lw=0.8, ls='--', zorder=1)

ax_tree.set_xlim(-0.5, n_leaves + 1.5)   # extra room for level labels
ax_tree.set_ylim(-0.3, L + 0.4)
ax_tree.axis('off')
ax_tree.set_title(r"Tree structure ($L=3$, $s=3$, 27 leaves)  —  entropy vs token position from start",
                  fontsize=11, pad=4)

# ---- Main loss plot ----
datasets = [
    ("v_8_L_3_seed_12345678_all_tokens_grand_avs.txt", "I-O"),
]

for filename, label in datasets:
    data = np.loadtxt(filename)
    t_keys, h_avg = weighted_avg_by(data, group_col=2)
    ax_main.plot(t_keys, h_avg, marker='o', markersize=3, linewidth=1.5,
                 label=label, color='#333333')

# Subtle coloured background per level-1 subtree
for s in range(b):
    ax_main.axvspan(s * (n_leaves // b) - 0.5, (s + 1) * (n_leaves // b) - 0.5,
                    alpha=0.07, color=palette[s], zorder=0)

ax_main.plot(t_keys,( np.log(inside_ell_1) /  np.log(v))*np.ones_like(t_keys), color='brown', label=r"$\mathcal{L}_{\ell=1}$", linestyle='--')

ax_main.plot(t_keys,( np.log(inside_ell_2) /  np.log(v))*np.ones_like(t_keys), color='purple', label=r"$\mathcal{L}_{\ell=2}$", linestyle='--')
ax_main.plot(t_keys,( np.log(inside_ell_3) /  np.log(v))*np.ones_like(t_keys), color='orange', label=r"$\mathcal{L}_{\ell=3}$", linestyle='--')

ax_main.set_xlabel(r"$t$ (position from start, 0 = first token)")
ax_main.set_ylabel(r"$H/\log(v)$")
ax_main.set_xlim(-0.4, 31)
ax_main.grid(True, alpha=0.3)
ax_main.legend()

plt.setp(ax_tree.get_xticklabels(), visible=False)
plt.tight_layout()
plt.show()
```
