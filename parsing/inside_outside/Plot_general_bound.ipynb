```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
v = 16
s = 3

plt.rcParams['legend.fontsize'] = 13  # or any size you prefer

# f range
f_vals = np.linspace(1e-4, 0.99, 1000)

# m = f * v
m_vals = f_vals * v**2

inside_ell_2 = (
    1
    + ((v - 1) / (v**s - 1))
    * (
        m_vals
        * (
            1
            + (v - 1) * (m_vals * v - 1) / (v**s - 1)
        )
        - 1
    )
)

inside_ell_3 = (
    1
    + ((v - 1) / (v**s - 1))
    * (
        m_vals
        * (
            inside_ell_2
        )
        - 1
    )
)

inside_ell_4 = (
    1
    + ((v - 1) / (v**s - 1))
    * (
        m_vals
        * (
            inside_ell_3
        )
        - 1
    )
)

inside_ell_inf = (v**s-v)/(v**s-1-m_vals*(v-1))

ell=2
# Expression inside the logarithm
# Lower bound
H_lower_ell_2 = np.log(inside_ell_2)
H_lower_ell_3 = np.log(inside_ell_3)
H_lower_ell_4= np.log(inside_ell_4)
H_lower_ell_inf = np.log(inside_ell_inf)

# Normalize by log(v)
H_norm_ell_2 = H_lower_ell_2 / np.log(v)
H_norm_ell_3 = H_lower_ell_3 / np.log(v)
H_norm_ell_4 = H_lower_ell_4 / np.log(v)
H_norm_ell_inf = H_lower_ell_inf / np.log(v)

# Saturate at 1
H_sat_ell_2 = np.minimum(H_norm_ell_2, 1.0)
H_sat_ell_3 = np.minimum(H_norm_ell_3, 1.0)
H_sat_ell_4 = np.minimum(H_norm_ell_4, 1.0)
H_sat_ell_inf = np.minimum(H_norm_ell_inf, 1.0)

data= np.loadtxt("v_16_L_2_range_grand_avs_1.txt")
f_data_L_2_1 = data[:-1, 0]
H_data_L_2_1 = data[:-1, 1]

data= np.loadtxt("v_16_L_2_range_grand_avs_2.txt")
f_data_L_2_2 = data[:, 0]
H_data_L_2_2 = data[:, 1]

data= np.loadtxt("v_16_L_3_range_grand_avs_1.txt")
f_data_L_3_1 = data[:-1, 0]
H_data_L_3_1 = data[:-1, 1]

data= np.loadtxt("v_16_L_3_range_grand_avs_2.txt")
f_data_L_3_2 = data[:, 0]
H_data_L_3_2 = data[:, 1]

data= np.loadtxt("v_16_L_4_range_grand_avs.txt")
f_data_L_4_1 = data[:, 0]
H_data_L_4_1 = data[:, 1]

# Plot
plt.figure(figsize=(6,4))

plt.plot(f_vals, H_sat_ell_2, linewidth=2,  label=r"$\mathcal{L}_{\ell=2}$")
plt.plot(f_vals, H_sat_ell_3, linewidth=2, label=r"$\mathcal{L}_{\ell=3}$")
plt.plot(f_vals, H_sat_ell_4, linewidth=2, label=r"$\mathcal{L}_{\ell=4}$")
plt.plot(f_vals, H_sat_ell_inf, linewidth=2, label=r"$\mathcal{L}_{\ell=\infty}$")
plt.scatter(np.concatenate([f_data_L_2_1, f_data_L_2_2]), np.concatenate([H_data_L_2_1, H_data_L_2_2]), s=30, zorder=5, label="I-O (L=2)")
plt.scatter(np.concatenate([f_data_L_3_1, f_data_L_3_2]), np.concatenate([H_data_L_3_1, H_data_L_3_2]), s=30, zorder=5, label="I-O (L=3)")
plt.scatter(f_data_L_4_1, H_data_L_4_1, s=30, zorder=5, label="I-O (L=4)")
# Horizontal saturation line
plt.axhline(
    1.0,
    linestyle="--",
    linewidth=1.5,
    #label=r"$\log(v)$ saturation"
)

plt.xlabel(r"$f$")
plt.ylabel(r"$H/\log(v)$")

plt.title("Last token conditional entropy (v=16, Varying tree RHM)")

plt.ylim(0, 1.05)

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```
